```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

data = [
    {"Name": "The Fellowship of the Ring", "Author": "John.R.R Tolkien", "Genre": "Fantasy", "Stars": 4, "type": "fiction"},
    {"Name": "The Two Towers", "Author": "John.R.R Tolkien", "Genre": "Fantasy", "Stars": 4, "type": "fiction"},
    {"Name": "The Return of the King", "Author": "John.R.R Tolkien", "Genre": "Fantasy", "Stars": 4, "type": "fiction"},
    {"Name": "The Hobbit", "Author": "J.R.R. Tolkien", "Genre": "Fantasy", "Stars": 5, "type": "fiction"},
    {"Name": "Dune", "Author": "Frank Herbert", "Genre": "Sci-Fi", "Stars": 5, "type": "fiction"},
    {"Name": "Dune Messiah", "Author": "Frank Herbert", "Genre": "Sci-Fi", "Stars": 4, "type": "fiction"},
    {"Name": "Children of Dune", "Author": "Frank Herbert", "Genre": "Sci-Fi", "Stars": 2, "type": "fiction"},
    {"Name": "The Player of Games", "Author": "Iain Banks", "Genre": "Sci-Fi", "Stars": 4, "type": "fiction"},
    {"Name": "Earthlight", "Author": "Arthur C. Clarke", "Genre": "Sci-Fi", "Stars": 3, "type": "fiction"},
    {"Name": "Foundation", "Author": "Isaac Asimov", "Genre": "Sci-Fi", "Stars": 5, "type": "fiction"},
    {"Name": "Second Foundation", "Author": "Isaac Asimov", "Genre": "Sci-Fi", "Stars": 4, "type": "fiction"},
    {"Name": "Foundation and Empire", "Author": "Isaac Asimov", "Genre": "Sci-Fi", "Stars": 3, "type": "fiction"},
    {"Name": "The Early Asimov - Volume 1", "Author": "Isaac Asimov", "Genre": "Sci-Fi", "Stars": 3, "type": "fiction"},
    {"Name": "Foundations Edge", "Author": "Isaac Asimov", "Genre": "Sci-Fi", "Stars": 3, "type": "fiction"},
    {"Name": "Fahrenheit 451", "Author": "Ray Bradbury", "Genre": "Sci-Fi", "Stars": 2, "type": "fiction"},
    {"Name": "Brave New World", "Author": "Aldous Huxley", "Genre": "Sci-Fi", "Stars": 4, "type": "fiction"},
    {"Name": "I, Robot", "Author": "Isaac Asimov", "Genre": "Sci-Fi", "Stars": 4, "type": "fiction"},
    {"Name": "Project Hail Mary", "Author": "_Andy Weir_", "Genre": "Sci-Fi", "Stars": 5, "type": "fiction"},
    {"Name": "Starship Troopers", "Author": "Robert A. Heinlein", "Genre": "Sci-Fi", "Stars": 4, "type": "fiction"},
    {"Name": "The Three Body Problem", "Author": "_Liu Cixin_", "Genre": "Sci-Fi", "Stars": 5, "type": "fiction"},
    {"Name": "The Dark Forest", "Author": "_Liu Cixin_", "Genre": "Sci-Fi", "Stars": 4, "type": "fiction"},
    {"Name": "Deaths End", "Author": "_Liu Cixin_", "Genre": "Sci-Fi", "Stars": 4, "type": "fiction"},
    {"Name": "Slaughterhouse-Five", "Author": "Kurt Vonnegut Jr.", "Genre": "Sci-Fi", "Stars": 5, "type": "fiction"},
    {"Name": "The Dispossessed", "Author": "Ursula K. Le Guin", "Genre": "Sci-Fi", "Stars": 5, "type": "fiction"},
    {"Name": "The Left Hand of Darkness", "Author": "Ursula K. Le Guin", "Genre": "Sci-Fi", "Stars": 4, "type": "fiction"},
    {"Name": "Revelation Space", "Author": "Alastair Reynolds", "Genre": "Sci-Fi", "Stars": 5, "type": "fiction"},
    {"Name": "Altered Carbon", "Author": "Richard K. Morgan", "Genre": "Sci-Fi", "Stars": 5, "type": "fiction"},
    {"Name": "Old Mans War", "Author": "John Scalzi", "Genre": "Sci-Fi", "Stars": 3, "type": "fiction"},
    {"Name": "Snow Crash", "Author": "Neal Stephenson", "Genre": "Sci-Fi", "Stars": 3, "type": "fiction", "year" : 2026},
    {"Name": "War of the Worlds", "Author": "H.G.Wells", "Genre": "Sci-Fi", "Stars": 2, "type": "fiction", "year" : 2026},
    {"Name": "Damascus Station", "Author": "David McCloskey", "Genre": "Spy", "Stars": 4, "type": "fiction"},
    {"Name": "The Spy Who Came in from the Cold", "Author": "John le Carré", "Genre": "Spy", "Stars": 5, "type": "fiction", "year" : 2026},
    {"Name": "Tinker Taylor Soldier Spy", "Author": "John le Carré", "Genre": "Spy", "Stars": 5, "type": "fiction", "year" : 2026},
    {"Name": "Where The Crawdads Sing", "Author": "Delia Owens", "Genre": "Other", "Stars": 5, "type": "fiction"},
    {"Name": "A Kestrel for a Knave", "Author": "Barry Hines", "Genre": "Other", "Stars": 4, "type": "fiction"},
    {"Name": "Life 3.0", "Author": "Max Tegmark", "Genre": "Science", "Stars": 4, "type": "non-fiction"},
    {"Name": "Regenesis: Feeding the World Without Devouring the Planet", "Author": "George Monbiot", "Genre": "Science", "Stars": 4, "type": "non-fiction"},
    {"Name": "The Order of Time", "Author": "Carlo Rovelli", "Genre": "Science", "Stars": 3, "type": "non-fiction"},
    {"Name": "The Rise and Fall of the Dinosaurs", "Author": "Steve Brusatte", "Genre": "Science", "Stars": 5, "type": "non-fiction"},
    {"Name": "How Civil Wars Start", "Author": "Barbara F. Walter", "Genre": "Politics", "Stars": 5, "type": "non-fiction"},
    {"Name": "How Westminster Works … and Why It Doesn’t", "Author": "Ian Dunt", "Genre": "Politics", "Stars": 5, "type": "non-fiction"},
    {"Name": "Politics on the Edge", "Author": "Rory Stewart", "Genre": "Politics", "Stars": 4, "type": "non-fiction"},
    {"Name": "Brexitland", "Author": "Maria Sobolewska and Robert Ford", "Genre": "Economics", "Stars": 5, "type": "non-fiction"},
    {"Name": "Economics: A User's Guide", "Author": "Ha-Joon Chang", "Genre": "Economics", "Stars": 5, "type": "non-fiction"},
    {"Name": "Utopia for Realists", "Author": "Rutger Bregman", "Genre": "Economics", "Stars": 3, "type": "non-fiction"},
    {"Name": "How to Be an Anticapitalist in the Twenty-First Century", "Author": "Erik Olin Wright", "Genre": "Economics", "Stars": 3, "type": "non-fiction"},
    {"Name": "Doughnut Economics", "Author": "Kate Raworth", "Genre": "Economics", "Stars": 4, "type": "non-fiction"},
    {"Name": "The Money Machine: How the City Works", "Author": "Philip Coggan", "Genre": "Economics", "Stars": 3, "type": "non-fiction"},
    {"Name": "The Anarchy: The Relentless Rise of the East India Company", "Author": "William Dalrymple", "Genre": "History", "Stars": 3, "type": "non-fiction"},
    {"Name": "Why Nations Fail: The origins of power, prosperity and poverty", "Author": "Daron Acemoglu & James A.Robinson", "Genre": "Economics", "Stars": 5, "type": "non-fiction", "year" : 2026},
    {"Name": "Contact", "Author": "Carl Sagan", "Genre": "Sci-Fi", "Stars": 4, "type": "fiction", "year" : 2026},
    {"Name": "Solaris", "Author": "Stanisław Lem", "Genre": "Sci-Fi", "Stars": 4, "type": "fiction", "year" : 2026},
    {"Name": "Shogun", "Author": "James Clavell", "Genre": "Historical Fiction", "Stars": 4, "type": "fiction", "year" : 2026},
    {"Name": "Children of Time", "Author": "Adrian Tchaikovsky", "Genre": "Sci-Fi", "Stars": 5, "type": "fiction", "year" : 2026},
    {"Name": "Homage to Catalonia", "Author": "George Orwell", "Genre": "Memoir", "Stars": 3, "type": "non-fiction", "year" : 2026},
    {"Name": "The Moon is a Harsh Mistress", "Author": "Robert A. Heinlein", "Genre": "Sci-Fi", "Stars": 3, "type": "fiction", "year" : 2026},
    {"Name": "The Constant Gardener", "Author": "John le Carré", "Genre": "Spy", "Stars": 4, "type": "fiction", "year" : 2026},
    {"Name": "Killers of the Flower Moon", "Author": "David Grann", "Genre": "History", "Stars": 4, "type": "non-fiction", "year" : 2026},
    {"Name": "Educated", "Author": "Tara Westover", "Genre": "Memoir", "Stars": 3, "type": "non-fiction", "year" : 2026},
]

df = pd.DataFrame(data)

# -----------------------------------
# Number of Stars plot
# -----------------------------------

star_counts = df["Stars"].value_counts().sort_index()
# print(star_counts)

plt.bar(star_counts.index, star_counts.values, color='orange', width=0.5)

plt.title("Book Ratings", fontsize=16)
plt.xlabel("Star Rating", fontsize=14)
plt.ylabel("Number of Books", fontsize=14)

plt.xticks([1,2,3,4,5])
plt.yticks(range(0, max(star_counts.values)+1,5))

plt.tick_params(axis='both', labelsize=12)

plt.savefig("Miscellaneous_books_star_rating_graph.png")
plt.show()

# -----------------------------------
# Number of fiction for each genre
# -----------------------------------

fiction_df = df[df["type"]=="fiction"]
genre_counts = fiction_df["Genre"].value_counts()
print(genre_counts)
plt.bar(genre_counts.index, genre_counts.values, color='orange', width=0.5)
plt.title("Number of Books per Genre (Fiction)", fontsize=16)
plt.xlabel("Genre", fontsize=14)
plt.ylabel("Number of Books", fontsize=14)

# Rotate labels so they don’t overlap
plt.xticks(rotation=10)

plt.yticks(range(0, max(genre_counts.values)+1, 5))
plt.tick_params(axis='both', labelsize=12)

plt.savefig("Miscellaneous_books_fiction_genre.png")
plt.show()

# -----------------------------------
# Number of non-fiction for each genre
# -----------------------------------

fiction_df = df[df["type"]=="non-fiction"]
genre_counts = fiction_df["Genre"].value_counts()

plt.bar(genre_counts.index, genre_counts.values, color='steelblue', width=0.5)
plt.title("Number of Books per Genre (Non-fiction)", fontsize=16)
plt.xlabel("Genre", fontsize=14)
plt.ylabel("Number of Books", fontsize=14)

# Rotate labels so they don’t overlap
plt.xticks(rotation=15)

plt.yticks(range(0, max(genre_counts.values)+1, 5))
plt.tick_params(axis='both', labelsize=12)

plt.savefig("Miscellaneous_books_non_fiction_genre.png")
plt.show()

# -----------------------------------
# Number of non-fiction for each genre
# -----------------------------------

genre_stats = (
    df.groupby("Genre")
      .agg(
          Average=("Stars", "mean"),
          Number=("Stars", "count"),
          Type=("type", "first")
      )
      .sort_values("Average")
)

# Blue for non-fiction, orange for fiction
colors = [
    "steelblue" if book_type == "non-fiction" else "orange"
    for book_type in genre_stats["Type"]
]

plt.figure(figsize=(10, 6))

bars = plt.bar(
    genre_stats.index,
    genre_stats["Average"],
    color=colors,
    width=0.6
)

# Add number of books above each bar
for bar, average, number in zip(
    bars,
    genre_stats["Average"],
    genre_stats["Number"]
):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        average + 0.05,
        f"{average:.2f}  (n={number})",
        ha="center",
        va="bottom",
        fontsize=11
    )

plt.title("Average Book Rating by Genre", fontsize=16)
plt.xlabel("Genre", fontsize=14)
plt.ylabel("Average Star Rating", fontsize=14)

plt.ylim(0, 5.5)
plt.yticks([1, 2, 3, 4, 5])

plt.tick_params(axis="both", labelsize=12)
plt.xticks(rotation=45, ha="right")

# Add legend for fiction / non-fiction
legend_elements = [
    Patch(facecolor="orange", label="Fiction"),
    Patch(facecolor="steelblue", label="Non-fiction")
]

plt.legend(
    handles=legend_elements,
    loc="upper center",
    bbox_to_anchor=(0.2, -0.3),
    ncol=2,
    frameon=False,
    fontsize=12
)

plt.tight_layout()
plt.savefig("Miscellaneous_books_average_rating_by_genre.png")
plt.show()
```
